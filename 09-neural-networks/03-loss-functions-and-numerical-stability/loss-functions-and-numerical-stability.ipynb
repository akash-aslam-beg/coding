```python
import numpy as np
```

```python
y_predicted = np.array([1,1,0,0,1])
y_true = np.array([0.30,0.7,1,0,0.5])
```

```python
def mae(y_predicted,y_true):
    total_error = 0
    for yp,yt in zip(y_predicted,y_true):
        total_error += abs(yp-yt)
    print('Total error is:{}'.format(total_error))
    mae = total_error/len(y_true)
    print('Mean absolute error is:{}'.format(mae))
    return mae
```

```python
mae(y_predicted,y_true)
```

```python
epsilon = 1e-15
```

```python
y_predicted_new = [max(i,epsilon) for i in y_predicted] 
y_predicted_new
```

```python
y_predicted_new = [min(i,1-epsilon) for i in y_predicted_new] 
y_predicted_new
```

```python
np.log(y_predicted_new)
```

```python
y = y_true
y_hat = y_predicted_new

y = np.array(y)
y_hat = np.array(y_hat)

-np.mean(y*np.log(y_hat)+(1-y)*np.log(1-y_hat))
```

```python
def log_loss(y, y_predicted):
    y_hat = [max(i,epsilon) for i in y_predicted]
    y_hat = [min(i,1-epsilon) for i in y_hat]
    y_hat = np.array(y_hat)
    return -np.mean(y*np.log(y_hat)+(1-y)*np.log(1-y_hat))
```

```python
log_loss(y, y_predicted)
```
