# Setting and resetting the index

```python
people = {
    "first": ["Corey", 'Jane', 'John'], 
    "last": ["Schafer", 'Doe', 'Doe'], 
    "email": ["person007@example.com", 'person014@example.com', 'person015@example.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['email']
```

```python
df.set_index('email')
```

```python
df
```

```python
df.set_index('email',inplace=True)
df
```

```python
df.index
```

```python
df.loc['person007@example.com']
```

```python
df.loc['person007@example.com','last']
```

```python
df.iloc[0]
```

```python
df.reset_index(inplace=True)
df
```
