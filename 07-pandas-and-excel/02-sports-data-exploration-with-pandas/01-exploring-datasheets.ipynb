```python
bro = pd.crosstab(df['Nation'],df['Team'])
brazilengland_row = bro.loc[['br BRA','eng ENG']]

brazilengland_row
```

```python
transposed_df = brazilengland_row.T
transposed_df.columns = ['Brazil', 'England']

transposed_df
```

```python
total_row = transposed_df.sum()

transposed_df.loc['Total'] = total_row

styled_df = transposed_df.style.set_properties(**{'font-weight': 'bold'}, subset=pd.IndexSlice['Total', :])

styled_df
```
