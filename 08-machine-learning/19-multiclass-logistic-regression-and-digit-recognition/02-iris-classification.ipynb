```python
df = pd.DataFrame(iris['data'],columns=iris['feature_names'])
df
```

```python
df['target'] = iris['target']
df
```

```python
df[df['target']==2].head()
```

```python
df['flower_name'] = df['target'].apply(lambda x: iris.target_names[x])
df
```

```python
pd.set_option('display.max_rows',150)
df
```

```python
df0 = df[df['target']==0]
df1 = df[df['target']==1]
df2 = df[df['target']==2]
```

```python
df2
```

```python
plt.figure(dpi=1000)
plt.style.use('seaborn')
plt.scatter(df0['sepal length (cm)'],df0['sepal width (cm)'],color='green',marker='o',edgecolor='black',linewidth=1,label='setosa (0)')
plt.scatter(df1['sepal length (cm)'],df1['sepal width (cm)'],color='blue',marker='o',edgecolor='black',linewidth=1,label='versicolor (1)')
plt.scatter(df2['sepal length (cm)'],df2['sepal width (cm)'],color='red',marker='o',edgecolor='black',linewidth=1,label='virginica (2)')
plt.legend()
plt.title('Iris Flower Sepal Measurement')
plt.xlabel('sepal length (cm)')
plt.ylabel('sepal width (cm)')
```

```python
plt.figure(dpi=1000)
plt.style.use('seaborn')
plt.scatter(df0['petal length (cm)'],df0['petal width (cm)'],color='green',marker='o',edgecolor='black',linewidth=1,label='setosa (0)')
plt.scatter(df1['petal length (cm)'],df1['petal width (cm)'],color='blue',marker='o',edgecolor='black',linewidth=1,label='versicolor (1)')
plt.scatter(df2['petal length (cm)'],df2['petal width (cm)'],color='red',marker='o',edgecolor='black',linewidth=1,label='virginica (2)')
plt.legend()
plt.title('Iris Flower Petal Measurement')
plt.xlabel('Petal length (cm)')
plt.ylabel('Petal width (cm)')
```

```python
inputs = df.drop(columns=['target','flower_name'])
inputs
```

```python
output = df['target']
output
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(inputs,output,train_size=0.8)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.predict(x_test)
```
