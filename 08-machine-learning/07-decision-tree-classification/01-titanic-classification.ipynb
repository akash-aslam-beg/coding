```python
import pandas as pd
```

```python
path = "data/titanic.csv"
```

```python
df = pd.read_csv(path)
df
```

```python
dfn = df[['Survived','Pclass','Sex','Age','Fare']]
dfn
```

```python
bro=pd.crosstab(df['Sex'],df['Survived'])
bro
```

```python
bro['Percentage_Survived'] = bro[1]/(bro[1]+bro[0])*100
bro['Percentage_Died'] = bro[0]/(bro[1]+bro[0])*100
bro
```

```python
fare = dfn.drop(columns='Sex')
fare.groupby('Survived').mean()
```

```python
inputs = dfn.drop(columns='Survived')
Target = dfn['Survived']
inputs
```

```python
from sklearn.preprocessing import LabelEncoder
le_sex = LabelEncoder()
```

```python
inputs['sex_n'] = le_sex.fit_transform(inputs['Sex'])
inputs = inputs.drop(columns='Sex')
inputs
```

```python
inputs = inputs[['Pclass', 'sex_n', 'Age', 'Fare']]

inputs
```

```python
from sklearn import tree
model = tree.DecisionTreeClassifier()
```

```python
model.fit(inputs,Target)
```

```python
model.score(inputs, Target)
```

```python
model.predict([[1,1,40,30]])
```

```python
model.predict_proba([[1,1,40,30]])
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(inputs,Target,train_size=0.9)
```
