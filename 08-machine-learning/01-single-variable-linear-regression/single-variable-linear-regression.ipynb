```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
from sklearn import linear_model
```

```python
df=pd.read_csv("data/homeprices (1).csv")
```

```python
df
```

```python
plt.scatter(df.area,df.price,color='red',marker='o')
plt.xlabel('area(sqr ft)')
plt.ylabel('price(US$)')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict([[3300]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
# y=mx+c
135.78767123*3300+180616.43835616432
```

```python
dfnew=pd.read_csv("data/prediction.csv")
```

```python
dfnew
```

```python
dfnew.drop(columns=['Unnamed: 0'],inplace=True)
```

```python
dfnew
```

```python
p=reg.predict(dfnew)
p
```

```python
dfnew['prices']=p
```

```python
dfnew
```

```python
dfnew.to_csv("data/prediction.csv",index=False)
```

```python
plt.scatter(df.area,df.price,color='red',marker='o')
plt.xlabel('area(sqr ft)')
plt.ylabel('price(US$)')
plt.plot(df.area,reg.predict(df[['area']]),color='green')
```

# Exercise

```python
canadadf=pd.read_csv("data/canada_per_capita_income.csv")
canadadf
```

```python
regc=linear_model.LinearRegression()
regc.fit(canadadf[['year']],canadadf['per capita income (US$)'])
```

```python
regc.predict([[2020]])
```

```python
years = np.array([[2017], [2018], [2019], [2020]])
r=regc.predict(years)
r
```

```python
predictdf = pd.DataFrame({
    'year': years.flatten(),
    'per capita income (US$)': r
})

newdf = canadadf._append(predictdf, ignore_index=True)
```

```python
newdf
```

```python
plt.scatter(canadadf.year,canadadf['per capita income (US$)'],color='red',marker='*')
plt.xlabel('Year')
plt.ylabel('per capita income (US$)')
plt.plot(canadadf.year,regc.predict(canadadf[['year']]),color='green')
```
