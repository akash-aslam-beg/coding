```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def logit(p):
    return np.log(p / (1 - p))

x = np.linspace(-10, 10, 400)

y_sigmoid = sigmoid(x)
p = np.linspace(0.01, 0.99, 400)  
y_logit = logit(p)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(x, y_sigmoid, label=r'$\sigma(x) = \frac{1}{1 + e^{-x}}$')
plt.title('Sigmoid Function')
plt.xlabel('x')
plt.ylabel(r'$\sigma(x)$')
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(p, y_logit, label=r'$\text{logit}(p) = \log\left(\frac{p}{1-p}\right)$'.replace(r'\text', ''), color='orange')
plt.title('Logit Function')
plt.xlabel('p')
plt.ylabel('logit(p)')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
