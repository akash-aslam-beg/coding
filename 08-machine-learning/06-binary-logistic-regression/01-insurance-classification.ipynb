```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
url = 'https://raw.githubusercontent.com/codebasics/py/master/ML/7_logistic_reg/insurance_data.csv'
df = pd.read_csv(url)
```

```python
df.head()
```

```python
plt.scatter(df['age'],df['bought_insurance'],color='red',edgecolor='black',linewidth=1,alpha=0.95)
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test=train_test_split(df[['age']],df['bought_insurance'],train_size=0.9)
```

```python
X_test
```

```python
X_train.head()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.predict(X_test)
```

```python
model.score(X_test,y_test)
```

```python
model.predict_proba(X_test)
```

```python
model.predict([[6]])
```

```python
m=model.coef_
m
```

```python
c=model.intercept_
c
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def predicition_function(age):
    z=m*age+c
    y=sigmoid(z)
    return y
```
