**Defining the Gaussian likelihood function**

```python
import numpy as np

def gaussian_likelihood(x, mean, std):
    """Calculate the Gaussian likelihood"""
    exponent = np.exp(-((x - mean) ** 2) / (2 * std ** 2))
    return (1 / (np.sqrt(2 * np.pi) * std)) * exponent
```

**Calculating the likelihood for each feature in df_0 and df_1**|

```python
df_0['P(Pclass)'] = gaussian_likelihood(df_0['Pclass'], df_0['Pclass'].mean(), df_0['Pclass'].std())
df_0['P(sex_n)'] = gaussian_likelihood(df_0['sex_n'], df_0['sex_n'].mean(), df_0['sex_n'].std())
df_0['P(Age)'] = gaussian_likelihood(df_0['Age'], df_0['Age'].mean(), df_0['Age'].std())
df_0['P(Fare)'] = gaussian_likelihood(df_0['Fare'], df_0['Fare'].mean(), df_0['Fare'].std())

df_1['P(Pclass)'] = gaussian_likelihood(df_1['Pclass'], df_1['Pclass'].mean(), df_1['Pclass'].std())
df_1['P(sex_n)'] = gaussian_likelihood(df_1['sex_n'], df_1['sex_n'].mean(), df_1['sex_n'].std())
df_1['P(Age)'] = gaussian_likelihood(df_1['Age'], df_1['Age'].mean(), df_1['Age'].std())
df_1['P(Fare)'] = gaussian_likelihood(df_1['Fare'], df_1['Fare'].mean(), df_1['Fare'].std())
```
