```python
import pandas as pd
```

```python
path = "data/titanic.csv"
df = pd.read_csv(path)
df
```

```python
dfn = df[['Survived','Pclass','Sex','Age','Fare']]
dfn
```

```python
inputs = dfn.drop(columns='Survived')
Target = dfn['Survived']
inputs
```

```python
from sklearn.preprocessing import LabelEncoder
le_sex = LabelEncoder()
```

```python
inputs['sex_n'] = le_sex.fit_transform(inputs['Sex'])
inputs = inputs.drop(columns='Sex')
inputs
```

```python
inputs = inputs[['Pclass', 'sex_n', 'Age', 'Fare']]

inputs
```

```python
inputs.columns[inputs.isna().any()]
```

```python
inputs.Age = inputs.Age.fillna(inputs.Age.mean())
inputs.head()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(inputs,Target,train_size=0.9)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
```

```python
model.fit(inputs,Target)
```

```python
model.score(inputs, Target)
```

```python
model.predict([[1,1,40,30]])
```

```python
model.predict_proba([[1,1,40,30]])
```

```python
model.score(x_test,y_test)
```

```python
model.predict(x_test)
```

```python
model.predict_proba(x_test[:10])
```

```python
from sklearn.model_selection import cross_val_score
cross_val_score(GaussianNB(),x_train, y_train, cv=5)
```
