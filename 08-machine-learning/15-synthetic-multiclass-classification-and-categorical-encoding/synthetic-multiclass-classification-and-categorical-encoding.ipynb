```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler

np.random.seed(42)

data = {
    'satisfaction_level': np.random.rand(1000),
    'average_montly_hours': np.random.randint(150, 300, size=1000),
    'promotion_last_5years': np.random.randint(0, 2, size=1000),
    'salary': np.random.choice(['low', 'medium', 'high'], size=1000),
    'Department': np.random.choice(['sales', 'technical', 'HR', 'accounting'], size=1000),
    'left': np.random.choice([0, 1, 2], size=1000)  
}

df = pd.DataFrame(data)

salary_dummies = pd.get_dummies(df['salary'], prefix="salary")
department_dummies = pd.get_dummies(df['Department'], prefix="Department")

df_with_dummies = pd.concat([df, salary_dummies, department_dummies], axis=1)
df_with_dummies.drop(columns=['salary', 'Department'], inplace=True)

X = df_with_dummies.drop(columns=['left'])
y = df['left']

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = LogisticRegression(multi_class='multinomial', solver='lbfgs', max_iter=1000)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")

print("Classification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
