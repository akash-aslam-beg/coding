```python
import pandas as pd

data = {
    'x': [0, 50, 100, 150, 200, 250, 300, 350, 400, 450, 500],
    'y': [5.94, 9.84, 9.84, 12.65, 14.68, 16.98, 18.90, 20.93, 22.49, 24.33, 25.96]
}

df = pd.DataFrame(data)

print(df)
```

```python
miu = df['x'].mean()
sigma = df['x'].std()

df_normalized = df.copy() 
df_normalized['x'] = (df['x'] - miu) / sigma

df_normalized
```

```python
x = df_normalized['x'].values
y = df_normalized['y'].values

def gradient_descent(x,y):
    m_curr=c_curr=0
    iterations=40000
    m=len(x)
    learning_rate=0.0001

    for i in range(iterations):
        y_predicted = m_curr*x + c_curr

        cost = (1/(2*m)) * np.sum((y_predicted - y) ** 2)

        c_d = (1/m) * np.sum(y_predicted - y)
        m_d = (1/m) * np.sum((y_predicted - y) * x)

        c_curr = c_curr - learning_rate * c_d
        m_curr = m_curr - learning_rate * m_d

        print('m={:<10.5f} c={:<10.5f} cost {:<10.5f} iteration {} '.format(m_curr,c_curr,cost, i))
    return m_curr,c_curr      

m,c=gradient_descent(x,y)
print('\nFinal Slope(m)={:<3.2f} and Intercept(c)={:3.2f}'.format(m,c))
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y, color='red', marker='o')
plt.plot(x, m * x + c , color='green')  
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Fit using Gradient Descent')
plt.show()

print('\nGradient Descent Slope(m)={:<3.2f} and Intercept(c)={:3.2f}'.format(m,c))
```

```python
import numpy as np
from sklearn import linear_model

x = df_normalized['x'].values.reshape(-1, 1)  
y = df_normalized['y'].values.reshape(-1, 1)  

reg = linear_model.LinearRegression()
reg.fit(x,y)

plt.scatter(x,y,color='red',marker='o')
plt.xlabel('x')
plt.ylabel('y')
plt.plot(x,reg.predict(x),color='green')

plt.show()

m_reg = reg.coef_[0]
c_reg=reg.intercept_
m_reg = reg.coef_[0][0]  
c_reg = reg.intercept_[0]

print('\nRegression Slope(m)={:<3.2f} and Intercept(c)={:3.2f}'.format(m_reg,c_reg))
```

```python
plt.scatter(x, y, color='red', marker='o', label='Data Points') 

plt.plot(x, m * x + c, color='green', label='Gradient Descent Line')

plt.plot(x, reg.predict(x), color='blue', label='Linear Regression Line')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Comparison of Gradient Descent and Linear Regression')
plt.legend()
plt.show()
```
