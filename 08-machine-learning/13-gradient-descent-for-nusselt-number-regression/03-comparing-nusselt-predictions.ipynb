```python
import pandas as pd
df_normalized=pd.read_csv("data/prediction_nusselt_normalized.csv")
```

```python
import numpy as np
from sklearn import linear_model
from sklearn.preprocessing import PolynomialFeatures
import matplotlib.pyplot as plt

X = df_normalized[['x1', 'x2', 'x3', 'x4', 'x5']].values  
y = df_normalized['nu'].values.reshape(-1, 1) 

degree = 3  
poly = PolynomialFeatures(degree)
X_poly = poly.fit_transform(X)

reg = linear_model.LinearRegression()
reg.fit(X_poly, y)

m_reg = reg.coef_
c_reg = reg.intercept_[0]

print('Polynomial Regression Coefficients:')
for i, coef in enumerate(m_reg[0], start=0):  
    print(f'Coefficient for feature {i} = {coef:<3.2f}')
print(f'Intercept(c) = {c_reg:3.2f}')

y_pred = reg.predict(X_poly)
```
