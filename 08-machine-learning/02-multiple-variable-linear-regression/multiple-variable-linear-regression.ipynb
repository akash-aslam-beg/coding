```python
hiringdf=pd.read_csv("data/hiring.csv")
hiringdf
```

```python
from word2number import w2n

hiringdf['experience'] = hiringdf['experience'].fillna('zero')

hiringdf['experience'] = hiringdf['experience'].apply(lambda value: w2n.word_to_num(value) if isinstance(value, str) else value)

median_test_score = round(hiringdf['test_score(out of 10)'].median())

hiringdf['test_score(out of 10)'] = hiringdf['test_score(out of 10)'].fillna(median_test_score)
```

```python
hiringdf
```

```python
regex = linear_model.LinearRegression()
regex.fit(hiringdf[['experience','test_score(out of 10)','interview_score(out of 10)']],hiringdf['salary($)'])
```

```python
regex.predict([[2,9,6]])
```

```python
regex.predict([[12,10,10]])
```
