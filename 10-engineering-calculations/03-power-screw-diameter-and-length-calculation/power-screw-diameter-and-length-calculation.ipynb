# Power-screw diameter and length calculation

Evaluating the coursework relation in SI units, then reporting diameters in mm and lengths in cm.

```python
import numpy as np

S_y = 390e6
T = 190.30

k = (np.pi * S_y) / (112 * T)
print(f'k = {k}')

def length(d):
    numerator = 112 * T * d
    denominator = (d**3 * np.pi * S_y) - (112 * T)
    if denominator == 0:
        return float('inf')  
    L = numerator / denominator
    return L

for i, d in enumerate(np.arange(20e-3, 0.8 * 64e-3, 0.8e-3)):
    L = length(d)
    L=100*L
    percentage = (100 * d / 64e-3)
    print(f'Iteration {i+1}: diameter = {d*1e3:.2f} mm, length = {L:.2f} centimeter, Percentage = {percentage:.2f}%')
```
