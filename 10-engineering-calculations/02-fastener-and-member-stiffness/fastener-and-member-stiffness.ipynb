# Fastener Stiffness

```python
Ad=0.1963
At=0.1419
ld=2.25
lt=0.75
E=30

k_b=(Ad*At*E)/(Ad*lt+At*ld)
print('Fastener Stiffness, k = {}'.format(k_b))
```

# Member Stiffness

```python
from numpy import *

alpha=(30*pi/180)
E=30
d=0.5
D=2.015
t=0.5

k=(pi*E*d*tan(alpha))/(log(((2*t*tan(alpha)+D-d)*(D+d))/((2*t*tan(alpha)+D+d)*(D-d))))

print('Member Stiffness, k = {}'.format(k))
```
