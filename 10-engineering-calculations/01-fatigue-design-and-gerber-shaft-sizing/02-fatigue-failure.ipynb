```python
import numpy as np

S_ut=690
S_e=234
S_y=580

S_m=(S_ut**2)/(2*S_e)*(1-(np.sqrt(1+(((2*S_e/S_ut)**2)*(1-(S_y/S_e))))))
print('S_m for Gerber:{}'.format(S_m))
```

```python
import numpy as np

r=1
S_ut=100
S_e=33.9

S_a=((r*S_ut)**2)/(2*S_e)*(-1+np.sqrt(1+((2*S_e)/(S_ut*r))**2))

print('S_a for Gerber:{}'.format(S_a))
```

```python
import numpy as np

S_ut=690 
S_e=234
sigma_a=51.6
sigma_m=51.6

n_f=0.5*((S_ut/sigma_m)**2)*(sigma_a/S_e)*(-1+np.sqrt(1+((2*sigma_a*S_e)/(S_ut*sigma_a))**2))

print('Gerber Factor of Safety:{:<4.2f}'.format(n_f))
```
