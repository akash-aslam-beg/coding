# Fatigue design calculations

Comparing four shaft-design relations using the same example loads. Moments and torques are in N·m; strengths are in Pa.

$$\sigma_a=\frac{16A}{\pi d^3},\quad \sigma_m=\frac{16B}{\pi d^3},\qquad A=\sqrt{4(K_fM_a)^2+3(K_{fs}T_a)^2},\quad B=\sqrt{4(K_fM_m)^2+3(K_{fs}T_m)^2}.$$

```python
import numpy as np
import pandas as pd

Se, Sut, Sy = 210e6, 700e6, 560e6
Kf, Kfs = 2.2, 1.8
Ma, Mm, Ta, Tm = 70, 55, 45, 35  # N·m
n = 2.0
A = np.sqrt(4*(Kf*Ma)**2 + 3*(Kfs*Ta)**2)
B = np.sqrt(4*(Kf*Mm)**2 + 3*(Kfs*Tm)**2)

goodman = 16*n/np.pi*(A/Se+B/Sut)
u, v = 16*n*A/(np.pi*Se), 16*n*B/(np.pi*Sut)
gerber = .5*(u+np.sqrt(u*u+4*v*v))
elliptic = 16*n/np.pi*np.sqrt((A/Se)**2+(B/Sy)**2)
soderberg = 16*n/np.pi*(A/Se+B/Sy)
diameters = np.cbrt([goodman, gerber, elliptic, soderberg])

sigma_a = 16*A/(np.pi*diameters**3)
sigma_m = 16*B/(np.pi*diameters**3)
checks = [n*(sigma_a[0]/Se+sigma_m[0]/Sut),
          n*sigma_a[1]/Se+(n*sigma_m[1]/Sut)**2,
          n*np.sqrt((sigma_a[2]/Se)**2+(sigma_m[2]/Sy)**2),
          n*(sigma_a[3]/Se+sigma_m[3]/Sy)]
assert np.allclose(checks, 1)
pd.DataFrame({'Criterion':['Goodman','Gerber','ASME elliptic','Soderberg'],
              'Diameter (mm)':diameters*1000})
```
