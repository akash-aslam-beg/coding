```python
from numpy import *

x = 0.5
y = 0.5

for i in range(1, 100):
    xnew = (1 / 10) * (3 * y * x ** 2 + 7)
    ynew = (1 / 5) * (y ** 2 + 4)

    if abs(xnew - x) < 0.000001 and abs(ynew - y) < 0.000001:
        break

    x = xnew  
    y = ynew
    print('Iteration:{} X={:.15f}     y={:.15f}'.format(i,x,y))

#print("\nSolution X={} y={}".format(xnew,ynew))
print("\nSolution X={:.5f} y={:.5f}".format(xnew, ynew))
```
