```python
import numpy as np
import matplotlib.pyplot as plt

def rk4_step(x, t, dt):

    k1 = f(x, t)
    k2 = f(x + 0.5 * dt * k1, t + 0.5 * dt)
    k3 = f(x + 0.5 * dt * k2, t + 0.5 * dt)
    k4 = f(x + dt * k3, t + dt)

    return x + (dt / 6) * (k1 + 2 * k2 + 2 * k3 + k4)

# f(x,t)=dx/dt = -x
def f(x, t):
    return -x

x0 = 1.0
t0 = 0.0
tf = 10.0
dt = 0.1

t_points = np.arange(t0, tf, dt)

x_points = np.zeros_like(t_points)
x_points[0] = x0

for i in range(1, len(t_points)):
    x_points[i] = rk4_step(x_points[i-1], t_points[i-1], dt) 

def exact_solution(t):
    return np.exp(-t)

exact_points = exact_solution(t_points)

plt.plot(t_points, x_points, color='green', label='RK4 Solution')
plt.plot(t_points, exact_points, 'rx', label='Exact Solution')

plt.xlabel('Time')
plt.ylabel('Value')
plt.title('Comparison of RK4 and Exact Solution')
plt.legend()
plt.grid(True)
plt.show()
```
