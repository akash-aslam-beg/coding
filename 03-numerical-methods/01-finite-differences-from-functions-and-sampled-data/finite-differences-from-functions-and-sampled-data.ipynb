# Using Function

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
dt=0.2
t = np.linspace(-2,4,60)
f= np.sin(t)
```

```python
dfdt= np.cos(t)
plt.figure(figsize=(15,5),dpi=100)
plt.plot(t,f,'k--',linewidth=1.2,label='Function')
plt.grid(True)
plt.plot(t,dfdt,'k',linewidth=3,label='Exact Derivative')
plt.legend()
plt.show()
```

```python
dfdtF=(np.sin(t+dt)-np.sin(t))/dt

dfdtB=(np.sin(t)-np.sin(t-dt))/dt

dfdtC=(np.sin(t+dt)-np.sin(t-dt))/(2*dt)
```

```python
plt.figure(figsize=(8,5),dpi=400)

plt.plot(t,dfdt,color='black',linewidth=1.2,label='Exact Derivative')
plt.plot(t,dfdtF,color='blue',linewidth=1.2,label='Forward Diff')
plt.plot(t,dfdtB,color='Green',linewidth=1.2,label='Backward Diff')
plt.plot(t,dfdtC,color='Violet',linewidth=1.2,label='Central Diff')

plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(5,2),dpi=400)

plt.plot(t,dfdt,color='black',linewidth=1.2,label='Exact Derivative')
plt.plot(t,dfdtF,color='blue',linewidth=1.2,label='Forward Diff')
plt.plot(t,dfdtB,color='Green',linewidth=1.2,label='Backward Diff')
plt.plot(t,dfdtC,color='violet',linewidth=1.2,label='Central Diff')

plt.legend()
plt.grid(True)
plt.xlim(0.85,1)
plt.show()
```

# Using Data sets instead of Function

```python
import numpy as np
from matplotlib import pyplot as plt

n=30
x=np.linspace(0.1,3,n)
f=np.sin(x)

plt.plot(x,f,'k')
plt.plot(x,f,'rx',linewidth=2)
plt.show()
```

```python
dx=x[1]-x[0]
dfdx=np.zeros(n)

dfdx[0]=(f[1]-f[0])/(x[1]-x[0])

for i in range(1,n-1):
    dfdx[i] = (f[i+1]-f[i-1])/(x[i+1]-x[i-1])

dfdx[-1]=(f[-1]-f[-2])/(x[-1]-x[-2])   

plt.figure(figsize=(8,5),dpi=400)
plt.plot(x,np.cos(x),color='black',label='True Derivative')
plt.plot(x,dfdx,'rx',label='Computed Derivative')

plt.legend()
plt.show()
```
