```python
# Given equation 2x^2-5x+3=0
#general formula x=x-(f(x))/differentiation of f(x)
x=7

for i in range(1,100):
    xnew = x-(2*x**2-5*x + 3)/(4*x-5)
    if abs(xnew-x)<0.000001:
        break
    x=xnew    
    print("Iteration:{} X={}".format(i,xnew))    

print("\nSolution X=%0.5f" %xnew)
```
