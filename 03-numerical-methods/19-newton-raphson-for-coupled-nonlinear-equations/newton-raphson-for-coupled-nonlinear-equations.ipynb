```python
import numpy as np

def f(xy):
    """
    System of nonlinear equations:
    3*y*x^2 - 10*x + 7 = 0
    y^2 - 5*y + 4 = 0
    """
    x, y = xy
    return np.array([
        3 * y * x**2 - 10 * x + 7,
        y**2 - 5 * y + 4
    ])

def J(xy):
    """
    Jacobian matrix of the system of equations
    """
    x, y = xy
    return np.array([
        [6 * y * x-10, 3 * x**2],
        [0, 2 * y - 5]
    ])

def newton_raphson_system(f, J, xy0, tol=1e-2, max_iter=100):

    xy = np.array(xy0, dtype=float)
    for iteration in range(max_iter):
        dxy = np.linalg.solve(J(xy), -f(xy))
        xy += dxy
        print(f"Iteration {iteration + 1}: x = {xy[0]:.6f}, y = {xy[1]:.6f}")
        if np.linalg.norm(dxy) < tol:
            return xy, iteration + 1
    raise ValueError("didn't converge")

initial_guess = [2, 1.5]
solution, iterations = newton_raphson_system(f, J, initial_guess)
print("Solution:", solution)
print("Iterations:", iterations)
```
