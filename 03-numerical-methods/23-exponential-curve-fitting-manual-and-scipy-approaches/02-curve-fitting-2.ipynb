```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from scipy import integrate as intg

plt.style.use('tableau-colorblind10')

data=pd.read_csv("data/graph2.csv")

x_value = data['x']
y_value = data['y']

def f(x,a,b):
    return a*(x**b)
a=0.5
b=2

fdata=f(x_value,a,b)

plt.scatter(x_value,y_value,s=100,edgecolor='black',linewidth=1,alpha=0.75,label='Experimental')
plt.plot(x_value,fdata,color='orange',label='Function')

#plt.title('y=ab^x Plot')
plt.title(r'$y = ax^{b}$ Plot',fontsize=15)
plt.legend()

TotalInt = intg.trapz(y_value,x_value)                     
TotalInt_func = intg.quad(f,0,5, args=(0.5,2))[0]        

print('Experimental Area:',TotalInt) 
print('Functional Area:',TotalInt_func)

plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit

print(plt.style.available)

plt.style.use('tableau-colorblind10')

data=pd.read_csv("data/graph.csv")
x_value = data['x']
y_value = data['y']

a=5
b=1.75664291

def f(x,a,b):
    return a*(b**x)

fdata=f(x_value,a,b)
plt.scatter(x_value,y_value,s=100,edgecolor='black',linewidth=1,alpha=0.75,label='Experimental Data')
plt.plot(x_value,fdata,color='green',label='Function')

#plt.title('y=ab^x Plot')
plt.title(r'$y = ab^{x}$ Plot Using Pandas',fontsize=15)
plt.legend()

popt, pcov = curve_fit(f,x_value,y_value,bounds=(1,5))
perr = np.sqrt(np.diag(pcov))

print(popt)
print(perr)

plt.show()
```
