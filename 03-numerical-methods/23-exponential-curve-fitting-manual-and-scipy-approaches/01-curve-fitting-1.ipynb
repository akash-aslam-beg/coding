```python
import numpy as np
from matplotlib import pyplot as plt

a=73.6998
b=1.1688
x=[1,2,3,4,5,6,7]

y=[87,97,113,129,202,195,193]

def f(x):
    return a*(b**x)

fdata=[f(xi) for xi in  x]
plt.scatter(x,y,s=100,edgecolor='black',linewidth=1,alpha=0.75)
plt.plot(x,fdata,color='orange')

#plt.title('y=ab^x Plot')
plt.title(r'$y = ab^{x}$ Plot',fontsize=15)

plt.show()
```
