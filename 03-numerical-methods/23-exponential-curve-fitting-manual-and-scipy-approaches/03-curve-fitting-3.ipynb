```python
import numpy as np
from matplotlib import pyplot as plt

c=0.541377
b=1.098
x=[1,2,3,4,5,6]

y=[1.50,4.60,13.90,40.10,125.10,299.50]

def f(x):
    return c*np.exp(b*x)

fdata=[f(xi) for xi in  x]
plt.scatter(x,y,s=100,edgecolor='black',linewidth=1,alpha=0.75)
plt.plot(x,fdata,color='orange')

#plt.title('y=ce^bx Plot')
plt.title(r'$y = ce^{bx}$ Plot',fontsize=15)

plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit
from scipy import integrate as intg

plt.style.use('seaborn')

c=0.541377
b=1.0698
# c=1
# b=1
x=np.linspace(1,6,6)

y=[1.50,4.60,13.90,40.10,125.10,299.50]

def f(x,b,c):
    return c*np.exp(b*x)

# a=0.5
# b=2

#fdata=[f(xi) for xi in  x_value]
fdata= f(x,b,c)

plt.figure(figsize=(10,5),dpi=120)
plt.scatter(x,y,color='yellow',s=100,edgecolor='black',linewidth=1,alpha=0.75,label='Experimental Data')
plt.plot(x,fdata,color='green',label='Function')

#plt.title('y=ab^x Plot')
plt.title(r'$y = ce^{bx}$ Plot',fontsize=15)
plt.legend()

popt, pcov = curve_fit(f,x,y,bounds=(1,6))
perr = np.sqrt(np.diag(pcov))

print(popt) 
print(perr)  

TotalInt = intg.trapz(y,x)                     
TotalInt_func = intg.quad(f,0,6, args=(b,c))[0]        
# low_Frac = intg.quad(func,0,2, args=(1,1.5369))[0]/TotalInt_func
# high_Frac = intg.quad(func,2,4, args=(1,1.5369))[0]/TotalInt_func

print('\nExperimental Area:',TotalInt) 
print('Functional Area:',TotalInt_func)

plt.show()
```
