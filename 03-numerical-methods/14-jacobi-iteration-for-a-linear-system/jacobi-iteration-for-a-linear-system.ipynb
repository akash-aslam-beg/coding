# Jacobi iteration for a linear system

The three unknowns are updated together using values from the previous iteration.

$$5x_1-2x_2+3x_3=-1,\quad -3x_1+9x_2+x_3=2,\quad -2x_1+x_2+7x_3=-3.$$

```python
import numpy as np

A = np.array([[5, -2, 3], [-3, 9, 1], [-2, 1, 7]], dtype=float)
b = np.array([-1, 2, -3], dtype=float)
x = np.zeros(3)
diagonal = np.diag(A)
remainder = A - np.diagflat(diagonal)

for iteration in range(1, 101):
    next_x = (b - remainder @ x) / diagonal
    change = np.linalg.norm(next_x - x, ord=np.inf)
    x = next_x
    if iteration <= 5 or change < 1e-10:
        print(f"{iteration:3d}: {x}")
    if change < 1e-10:
        break

print(f"Residual: {np.linalg.norm(A @ x - b, ord=np.inf):.2e}")
assert np.allclose(A @ x, b, atol=1e-8)
```
