```python
x2=0
x3=0

for i in range(1,10):
    x1=(-1/5)+(2/5)*x2-(3/5)*x3
    x2=(2/9)+(3/9)*x1-(1/9)*x3
    x3=(-3/7)+(2/7)*x1-(1/7)*x2

    print('Iteration {}: x1={}   x2={}    x3={}'.format(i,round(x1,3),round(x2,3),round(x3,3)))
```

```python
x2=0

for i in range(1,10):
    x1=(6+x2)/7
    x2=(x1+4)/5

    print('Iteration {}: x1={}  x2={}'.format(i,round(x1,3),round(x2,3)))
```

```python
y=0
z=0

for i in range(1,10):
    x=(3+z-y)/4
    y=(19-2*x-z)/7
    z=(31+3*y-x)/12

    print('Iteration {}: x={}   y={}    z={}'.format(i,round(x,3),round(y,3),round(z,3)))
```
