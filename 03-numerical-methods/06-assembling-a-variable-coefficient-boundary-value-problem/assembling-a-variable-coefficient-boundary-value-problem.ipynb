# Assembling a variable-coefficient boundary-value problem

The coefficients are evaluated at each interior node.

$$y''+xy'-xy=2x,\qquad y(0)=1,\quad y(2)=8.$$ 

```python
import numpy as np
import pandas as pd

x = np.linspace(0, 2, 5)
h = x[1]-x[0]
A = np.zeros((len(x), len(x)))
b = 2*x*h*h
A[0, 0] = A[-1, -1] = 1
b[0], b[-1] = 1, 8
for i in range(1, len(x)-1):
    A[i, i-1:i+2] = [1-x[i]*h/2, -2-x[i]*h*h, 1+x[i]*h/2]
y = np.linalg.solve(A, b)
assert np.allclose(A @ y, b)
pd.DataFrame({'x': x, 'y': y})
```
