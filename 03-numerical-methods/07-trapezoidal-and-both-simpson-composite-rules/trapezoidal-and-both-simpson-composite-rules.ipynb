```python
from numpy import *
def f(x):
    return x**3

a=0
b=2
n=30
h=(b-a)/n
summation=0
for i in range(1,n):
    summation=summation+ h*f(a+i*h)

I = (h/2)*(f(a)+f(b)) + summation

print("Integration Result = {}".format(I))
```

```python
a= 0
b= 2
n= 20

def f(x):
    return x**3

def simpson13(a,b,n):
    h= (b-a)/n
    sum= f(a)+f(b)

    for i in range(1,n):
        c = a+i*h
        if i%2==0:
            sum+=2*f(c)
        else:
            sum+=4*f(c)

    ans= (h/3)*sum
    return ans

result= simpson13(a,b,n)
print("Integration result by Simpson 1/3rd rule is: ",result)
```
