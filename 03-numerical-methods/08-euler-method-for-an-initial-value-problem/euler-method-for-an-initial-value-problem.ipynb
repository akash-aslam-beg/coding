```python
def f(x, y):
    return y - x**2 + 1

x = 0
y = 0.5
h = 0.25
n = 15

for i in range(1, n+1):
    m1 = f(x, y)
    m2 = f(x + h, y + h * m1)
    yn = y + h * (m1 + m2) / 2
    xn = x + h

    print('x{} = {:.2f}, y{} = {:.6f}'.format(i, xn, i, yn))

    y = yn
    x = xn
```
