```python
from scipy.optimize import newton,bisect,fsolve,root

f = lambda x:2*x**2-5*x+3

print('By Newton Raphson Method X={}'.format(newton(f,5)))
print('By Bisection Method X={}'.format(bisect(f,1.4,9)))
print('By using fsolve X={}'.format(fsolve(f,4)))

g=[5,1,0,13,3,1.4]
print('By using fsolve X={}'.format(fsolve(f,g)))
```
