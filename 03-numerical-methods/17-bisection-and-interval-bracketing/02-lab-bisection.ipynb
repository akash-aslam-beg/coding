```python
#Given Equation : x^3-x-1=0

x1=0
x2=5

def f(a):
    value = a**3 - a -1
    return value

if (f(x1)*f(x2))>0:
    print("The root is not available within the interval since sign is same")
    exit

for i in range(1, 100, 1):

    xh = (x1 + x2) / 2.0 
    print('Iteration {}:   {}\t\terror: {}'.format(i,xh,(abs(xh-x1)/xh)*100))
    print('')

    if abs(f(x1))<1.0E-6: 
        break

    elif (f(xh) * f(x1)) <0: 
        x2 = xh   

    else :
        x1 = xh   

print("\nSolution of the equation is X = {}".format(round(x1,8)))   
print("Total bisection done: {}".format(i))
```
