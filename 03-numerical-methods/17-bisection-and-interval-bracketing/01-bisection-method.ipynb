```python
x1=0
x2=1.3

def f(a):
    value = 2*a**2 - 5*a + 3
    return value

#or y=lambda x: 2*x**2-5*x+3

if (f(x1)*f(x2))>0:
    print("The root is not available within the interval")
    exit

for i in range(1, 100, 1):

    xh = (x1 + x2) / 2.0 

    if abs(f(x1))<1.0E-6: 
        break

    elif (f(xh) * f(x1)) <0: 
        x2 = xh   

    else :
        x1 = xh   

print("Solution of the equation is X = {}".format(round(x1,4)))   
print("Total bisection done: {}".format(i))
```
