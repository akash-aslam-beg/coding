```python
import numpy as akash
import matplotlib.pyplot as graph 

time=akash.arange(0,1000,0.05)
sine=akash.sin(time*akash.pi/180)

graph.figure(figsize=(15,5),dpi=100) 

graph.plot(time,sine,lw=2,color='red',label='sine') 

graph.legend(loc='upper left') #for position of legend #legend(loc=(0.7,0.5)) to modify position

graph.title("Sine Function",color='Blue',fontsize='25')

graph.xlabel('Time',color='green',fontsize=18)
graph.ylabel('Sine',color='green',fontsize=18)

graph.grid()

#graph.xlim(time[0],time[-1])

graph.show()
```

```python
import numpy as akash
import matplotlib.pyplot as graph

x=akash.arange(0,20,1)
y1=2*x**2
y2=x**2-3*x
graph.plot(x,y1,color='blue',label='y1',marker='*')
graph.plot(x,y2,color='green',label='y2',marker='*')
graph.legend()
graph.legend(loc=(.5,.5))
graph.grid()

graph.show()
```

# Multiple Plots with different axes

```python
import numpy as akash
import matplotlib.pyplot as graph

x=akash.arange(0,20,1)
y1=2*x**2
y2=x**2-3*x

newfig=graph.figure()
axes1=newfig.add_axes([0,0,2,1.15])
axes1=graph.plot(x,y1,color='blue',marker='*')

axes2=newfig.add_axes([0.30,0.6,0.4,0.4])
axes2=graph.plot(x,y2,color='green',marker='*')

axes3=newfig.add_axes([0.8,0.6,0.4,0.4])
axes3=graph.plot(x,y1,color='red',marker='*')

graph.show()
```
