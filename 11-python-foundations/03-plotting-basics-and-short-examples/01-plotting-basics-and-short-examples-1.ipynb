```python
import numpy as akash
import matplotlib.pyplot as graph

x=akash.arange(0,20,1)
y1=2*x**2
y2=x**2-3*x

newfig=graph.figure(figsize=(4,4))
axes1=newfig.add_axes([0,0,2,1.15])
axes1=graph.plot(x,y1,color='blue',marker='*')

axes2=newfig.add_axes([0.30,0.6,0.4,0.4])
axes2=graph.plot(x,y2,color='green',marker='*')
graph.title('yoyo',color='green')

axes3=newfig.add_axes([0.8,0.6,0.4,0.4])
axes3=graph.plot(x,y1,color='red',marker='*')

graph.show()
```

```python
from numpy import *               
from matplotlib.pyplot import *

theta=arange(0,1000,0.001)
y1=sin(theta*pi/180)
y2=cos(theta*pi/180)
y3=tan(theta*pi/180)

figure(figsize=(15,6))

subplot(131)
plot(theta,y1,color='red')
#title('Sine',fontsize='30',color='red')    
title('Sin'r'($\theta{}$)',fontsize='30',color='red') 

subplot(132)
plot(theta,y2,color='blue')
title('Cos'r'($\theta{}$)',fontsize='30',color='blue')

subplot(133)
plot(theta,y3,color='green')
title('tan'r'($\theta{}$)',fontsize='30',color='green')

tight_layout(pad=4)
```
