```python
from numpy import *
x=0.91020796
y=0.99519231

Xk=array([[x],[y]]) 
print('Initital guess Matrix')
print(Xk)

J=array([[6*x*y-10,3*x**2],[0,2*y-5]]) 
print('\nJacobian Matrix')
print(J)

J_inverse = linalg.inv(J) 

print('\nInverse of the Jacobian Matrix')
print(J_inverse)

Fk=array([[3*y*x**2-10*x+7],[y**2-5*y+4]]) 
print('\nMatrix of the functions')
print(Fk)

multiplication = J_inverse @ Fk
print('\nmultiplication of J inverse and Fk')
print(multiplication)

Xk1=Xk-multiplication
print('\nFinal value: Xk - Jinverse into F')
print(Xk1)
```

```python
import numpy as np

x=3.2122
y=2.3176

Xk=array([[x],[y]]) 
print('Initital guess Matrix')
print(Xk)

J=array([[(-0.25*(np.exp(x/2)+np.exp(-x/2))),1],[18*x,50*y]]) 
print('\nJacobian Matrix')
print(J)

J_inverse = linalg.inv(J) 

print('\nInverse of the Jacobian Matrix')
print(J_inverse)

Fk=array([[y-0.5*(np.exp(x/2)-np.exp(-x/2))],[9*x**2+25*y**2-225]]) 
print('\nMatrix of the functions')
print(Fk)

multiplication = J_inverse @ Fk
print('\nmultiplication of J inverse and Fk')
print(multiplication)

Xk1=Xk-multiplication
print('\nFinal value: Xk - Jinverse into F')
print(Xk1)
```
