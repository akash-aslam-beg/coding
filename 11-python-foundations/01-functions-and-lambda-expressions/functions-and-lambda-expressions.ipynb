# Functions and lambda expressions

A volume calculation and a polynomial evaluation from the introductory Python exercises.

```python
def volume(length, width, height):
    return length * width * height

polynomial = lambda x: 2*x**2 - 5*x + 3
print(f"Cuboid volume: {volume(2, 3, 4)} cubic units")
print(f"Polynomial at x = 30: {polynomial(30)}")
```
