# For Loop

```python
for i in range(1,11,1): 
    i=2*i
    print(i)
print('\n')    
print(i)
```

```python
import numpy as np 
y=np.linspace(3,18,6) 
x=np.linspace(1,10,10)
print(y)
print(x)
```

```python
x = [1, 2, 3, 4, 5]
for i in x:
    product=2*i           
    print(product)
```

```python
x = [1, 2, 3, 4, 5]

y=[]    

for i in x:
    product=2*i           
    y.append(product)
    
print(y)  
print('The 4th number is {}'.format(y[3]))
print('The 4th number is ',y[3])
```

```python
n= int(input("Enter the number till which you want thte sum  ")) 

sum=0
for i in range(1,n+1,1):
    sum+=i

print('\n')
print("sum is {}".format(sum))  
print("sum is ",sum)
```
