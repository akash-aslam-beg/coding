```python
fruit_colors = {
    'apple': 'red',
    'banana': 'yellow',
    'grape': 'purple'
}

for fruit, color in fruit_colors.items():
    print('The color of {} is {}'.format(fruit,color))
```

```python
students = {
    'Alice': {'age': 25, 'major': 'Physics'},
    'Bob': {'age': 23, 'major': 'Mathematics'},
    'Charlie': {'age': 24, 'major': 'Chemistry'}
}

for name, details in students.items():
        print('{} is {} years old and major in {}'.format(name,details['age'],details['major']))
```

```python
team_scores = {
    'Team A': [10, 15, 20],
    'Team B': [12, 18, 22],
    'Team C': [14, 19, 25]
}

for team, scores in team_scores.items():
    avg_score = sum(scores) / len(scores)
    print('{} has an average score of {}'.format(team,avg_score))
```

```python
animals = ['cat', 'dog', 'rabbit', 'horse']

for index, animal in enumerate(animals):
    print('Index {}: {}'.format(index, animal))
```

```python
for index, animal in enumerate(animals, start=1):
    print('Index {}: {}'.format(index, animal))
```

```python
names = ['Alice', 'Bob', 'Charlie']
scores = [85, 92, 88]

for name, score in zip(names, scores):
    print('{} scored {} marks.'.format(name, score))
```

```python
names = ['Alice', 'Bob', 'Charlie']
scores = [85, 92]

for name, score in zip(names, scores):
    print('{} scored {} marks.'.format(name, score))
```

```python
names = ['Alice', 'Bob', 'Charlie']
scores = [85, 92, 88]

for index, (name, score) in enumerate(zip(names, scores)):
    print('Student {}: {} scored {} marks.'.format(index + 1, name, score))
```
