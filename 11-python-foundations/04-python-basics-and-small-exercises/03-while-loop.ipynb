# while loop

```python
sum=0
i=1
n=int(input("Enter the value "))

while i<=n:
    sum+=i
    i+=1

print("The sum of 1 to {} is {}".format(n,sum))
```

```python
sum=0
i=1
n=int(input("Enter the value "))

while i<=n:
    sum+=i
    i+=1
    print('i= ',i-1,'sum is',sum)

print("\n\nThe sum of 1 to {} is {}".format(n,sum))
```
