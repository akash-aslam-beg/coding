```python
student = {
    "name": "Alice",
    "age": 20,
    "major": "Computer Science",
    "GPA": 3.8
}

print("Student's name:", student["name"])  
print("Student's age:", student["age"])    
print("Student's major:", student["major"])  

student["age"] = 21
print("Student's updated age:", student["age"])  

student["year"] = "Junior"
print("Student's year:", student["year"])  

del student["GPA"]
print("Student dictionary after removing GPA:", student)

print("Student details:")
for key, value in student.items():
    print(key + ":", value)
```

# Counter



```python
from collections import Counter
c=Counter(['Python','Javascript'])
print(c)
c=Counter(['Python','Javascript','Akash','Akash'])
print(c)
c.update(['C++','Akash'])
print(c)
```
