```python
import numpy as np

T1 = 200.0
T2 = 200.0

T1_list = [T1]
T2_list = [T2]

num_steps = 41

A = np.array([[-1.9375, 0.46875], 
              [0.9375, -1.96767]])

time_step = 15

for i in range(1, num_steps):
    
    prev_T1 = T1_list[-1]
    prev_T2 = T2_list[-1]

    B = np.array([-prev_T1 - 33.482, 
                  -prev_T2 - 34.386])

    T_new = np.linalg.solve(A, B)

    T1_list.append(T_new[0])
    T2_list.append(T_new[1])

    time = i * time_step
    print('Iteration: {:<5} time: {:<5.0f} T1: {:<8.2f} T2: {:<8.2f}'.format(i, time, T_new[0], T_new[1]))
```
