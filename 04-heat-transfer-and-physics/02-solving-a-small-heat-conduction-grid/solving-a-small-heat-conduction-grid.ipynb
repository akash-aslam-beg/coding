```python
import numpy as np

nx, ny = 3, 3

top_temp = 0
left_temp = 100
bottom_temp = 100
right_temp = 100

N = nx * ny

A = np.zeros((N, N))
B = np.zeros(N)

def idx(i, j):
    return i * nx + j

for i in range(ny):
    for j in range(nx):
        k = idx(i, j)
        if i == 0:  # Top boundary
            A[k, k] = 1
            B[k] = top_temp
        elif i == ny-1:  # Bottom boundary
            A[k, k] = 1
            B[k] = bottom_temp
        elif j == 0:  # Left boundary
            A[k, k] = 1
            B[k] = left_temp
        elif j == nx-1:  # Right boundary
            A[k, k] = 1
            B[k] = right_temp
        else:  
            A[k, k] = -4
            A[k, idx(i+1, j)] = 1
            A[k, idx(i-1, j)] = 1
            A[k, idx(i, j+1)] = 1
            A[k, idx(i, j-1)] = 1

#Solve the system AY = B
Y = np.linalg.solve(A, B)

T = Y.reshape((ny, nx))

print("Temperature distribution in the grid:")
print(T)
assert np.isclose(T[1,1], 75)
```
