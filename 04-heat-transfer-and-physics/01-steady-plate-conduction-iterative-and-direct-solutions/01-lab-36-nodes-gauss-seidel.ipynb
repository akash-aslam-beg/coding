```python
import numpy as np

nx, ny = 6, 6

top_temp = 0
left_temp = 100
bottom_temp = 400
right_temp = 200

# Initialize temperature grid with boundary conditions
T = np.zeros((ny, nx))

# Set boundary conditions
T[0, :] = top_temp         # Top boundary
T[:, 0] = left_temp        # Left boundary
T[-1, :] = bottom_temp     # Bottom boundary
T[:, -1] = right_temp      # Right boundary

max_iter = 10000
tolerance = 1e-6

for iteration in range(max_iter):
    T_old = T.copy()
    for i in range(1, ny-1):
        for j in range(1, nx-1):
            T[i, j] = 0.25 * (T[i+1, j] + T[i-1, j] + T[i, j+1] + T[i, j-1])

    diff = np.max(np.abs(T - T_old))
    if diff < tolerance:
        break

print("Temperature distribution in the grid after", iteration, "iterations:")
print(T)
```
