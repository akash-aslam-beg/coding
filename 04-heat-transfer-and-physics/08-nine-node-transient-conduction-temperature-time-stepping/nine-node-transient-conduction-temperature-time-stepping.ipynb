```python
import numpy as np

delta_t = 10  
total_time = 60 * 60  
steps = total_time // delta_t

T = np.full((9, steps + 1), 90.0)  

a1 = 0.0836
a2 = 0.222
a3 = 0.444
c1 = 11.2
c2 = 22.4
c3 = 12.8
c4 = 109.2
c5 = 424
c6 = 202.4
c7 = 105.2

for i in range(steps):
    T[0, i + 1] = a1 * T[0, i] + a3 * (T[1, i] + T[3, i]) + a3 * c1
    T[1, i + 1] = a1 * T[1, i] + a2 * (T[0, i] + T[2, i] + 2 * T[4, i]) + a2 * c2
    T[2, i + 1] = 0.0552 * T[2, i] + a3 * (T[1, i] + T[5, i]) + a3 * c3
    T[3, i + 1] = 0.112 * T[3, i] + a2 * (T[0, i] + 2 * T[4, i]) + a2 * c4
    T[4, i + 1] = 0.112 * T[4, i] + a2 * (T[1, i] + T[3, i] + T[5, i] + T[7, i]) + a2 * c4
    T[5, i + 1] = 0.0931 * T[5, i] + 0.074 * (T[2, i] + 4 * T[4, i] + 2 * T[8, i]) + 0.074 * c5
    T[6, i + 1] = a1 * T[6, i] + a2 * (T[3, i] + T[7, i]) + a2 * c6
    T[7, i + 1] = a1 * T[7, i] + a2 * (T[4, i] + T[6, i] + T[8, i]) + a2 * c6
    T[8, i + 1] = a1 * T[8, i] + a3 * (T[5, i] + T[7, i]) + a3 * c7

print("Temperature at Node 3:")
print(f"At 1 minute: {T[2, 1 * 60 // delta_t]:.1f}°C")
print(f"At 3 minutes: {T[2, 3 * 60 // delta_t]:.1f}°C")
print(f"At 5 minutes: {T[2, 5 * 60 // delta_t]:.1f}°C")
print(f"At 10 minutes: {T[2, 10 * 60 // delta_t]:.1f}°C")
print(f"At 60 minutes: {T[2, 60 * 60 // delta_t]:.1f}°C")
```
