```python
import numpy as akash
import matplotlib.pyplot as graph
from mpl_toolkits.mplot3d import axes3d

x=akash.arange(-5,5,0.1)
y=akash.arange(-5,5,0.1)

X,Y=akash.meshgrid(x,y)
Z=(X**2+Y**2)

graph.figure(figsize=(7,5))

axes=graph.axes(projection='3d') 
mysurface=axes.plot_surface(X,Y,Z,cmap='hot')

cset=graph.contour(X,Y,Z,zdir='z',offset=0,cmap='jet') 

axes.set_xlabel('Value of X')
axes.set_ylabel('Value of Y')

graph.colorbar(mysurface)
graph.show()
```
