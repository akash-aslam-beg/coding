# Plotting a curve in polar coordinates

The radial relation is evaluated as an array so the samples form one continuous curve.

$$r(\theta)=\frac{ab}{|a\sin\theta|+|b\cos\theta|},\qquad a=7,\quad b=4.$$

```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 7, 4
theta = np.linspace(0, 2*np.pi, 721)
r = a*b/(np.abs(a*np.sin(theta)) + np.abs(b*np.cos(theta)))
fig, ax = plt.subplots(figsize=(5, 5), subplot_kw={'projection':'polar'})
ax.plot(theta, r, color='#0f766e', linewidth=2)
ax.set_title('Polar-coordinate curve', pad=20)
plt.tight_layout()
plt.show()
assert np.all(np.isfinite(r)) and np.isclose(r[0], a)
```
