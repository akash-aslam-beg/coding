```python
import numpy as akash
import matplotlib.pyplot as graph

xp=akash.linspace(-5,5,5)
yp=akash.linspace(-5,5,5)

#xp=akash.arange(-50,50,10)
#yp=akash.arange(-50,50,10)

X,Y= akash.meshgrid(xp,yp)

Z=akash.sin(X)*akash.cos(Y)

graph.contour(X,Y,Z,50,cmap='jet') 
graph.colorbar()

graph.show()
```
