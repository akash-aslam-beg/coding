```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')

sns.set(style="ticks")
sns.pairplot(iris, hue="species", markers=["o", "s", "D"])

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')

sns.set(style="whitegrid")
g = sns.pairplot(iris, hue="species", palette="husl", markers=["o", "s", "D"], height=2.5)

g.fig.suptitle("Pair Plot of Iris Dataset", y=1.02)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

sns.set(style="darkgrid")

sns.scatterplot(x="total_bill", y="tip", hue="sex", palette="viridis", data=tips)
plt.title("Scatter Plot with Viridis Palette")
plt.show()

sns.barplot(x="day", y="total_bill", hue="sex", palette="pastel", data=tips)
plt.title("Bar Plot with Pastel Palette")
plt.show()

sns.violinplot(x="day", y="total_bill", hue="sex", palette="deep", data=tips, split=True)
plt.title("Violin Plot with Deep Palette")
plt.show()

sns.boxplot(x="day", y="total_bill", hue="sex", palette="colorblind", data=tips)
plt.title("Box Plot with Colorblind Palette")
plt.show()
```
